```python
import pandas as pd
```

```python
# create a new dataframe(2 dimensional arr)
df = pd.read_csv("credit_risk_dataset.csv")
```

```python
# inspect the dataset
print(df.info())
print("\n")
print(df.describe())
```

```python
# view the mean of all columns

for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        print(f"{col} mean",df[col].mean())
        print(f"{col} median ", df[col].median())
        print(f"{col} mode", df[col].mode())
```

```python
# show the standard deviation aka how much does the values differ from the avg value.
# view the mean of all columns

for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        print(f"{col}",df[col].std())
```

```python
# show the range of values(visually say which ones are categorical/numerical features)
for col in df.columns:
    print(f"{col}", df[col].value_counts())
```
