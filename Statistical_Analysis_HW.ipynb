## Column-by-column review

The dataset has numerical and categorical columns. The descriptive statistics above cover numerical columns; the tables below summarize every categorical column with counts and proportions. Missing values are also reported so the summaries can be interpreted in context.


```python
categorical_columns = credit_risk_data.select_dtypes(exclude="number").columns

categorical_summary = pd.concat(
    {
        column: pd.DataFrame({
            "count": credit_risk_data[column].value_counts(dropna=False),
            "proportion": credit_risk_data[column].value_counts(dropna=False, normalize=True),
        })
        for column in categorical_columns
    },
    names=["column", "category"],
)
categorical_summary
```

```python
column_quality = pd.DataFrame({
    "dtype": credit_risk_data.dtypes.astype(str),
    "missing_count": credit_risk_data.isna().sum(),
    "missing_percent": credit_risk_data.isna().mean().mul(100).round(2),
    "unique_values": credit_risk_data.nunique(dropna=False),
})
column_quality
```

## Absolute correlation with the target

The dataset target is `loan_status`. The next cell ranks numeric columns by the absolute value of their Pearson correlation with that target. The target itself is removed from the ranking.


```python
target_column = "loan_status"
target_correlations = (
    credit_risk_data.corr(numeric_only=True)[target_column]
    .drop(target_column)
    .sort_values(key=lambda correlations: correlations.abs(), ascending=False)
)
target_correlations.to_frame(name="correlation_with_loan_status")
```

### Interpretation

The strongest numerical associations in this dataset are `loan_percent_income` (about **0.379**) and `loan_int_rate` (about **0.335**). Both are positive: higher values tend to occur alongside `loan_status = 1`. If `1` denotes a default, the first relationship is plausible because a larger loan burden relative to income can make repayment harder. The interest rate may also reflect risk pricing: lenders can charge higher rates to borrowers they assess as riskier, and higher rates also increase the repayment burden.


These explanations are consistent with empirical research on consumer and peer-to-peer lending. For example, a study of LendingClub loans found that borrower indebtedness and interest rates were associated with default risk ([Duarte, Siegel & Young, 2015](https://journals.plos.org/plosone/article?id=10.1371/journal.pone.0139427)). That study uses a different lending platform and dataset, so it provides context rather than proof about this dataset.


Pearson correlation measures linear association, not causation. The interest rate may partly encode a lender’s existing risk assessment, and the features may be related to one another. Correlation also covers numeric columns only; it does not measure the relationship of the categorical columns to the target.
