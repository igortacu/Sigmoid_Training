```python
import pandas as pd
```

```python
# create a new dataframe(2 dimensional arr)
df = pd.read_csv('credit_risk_dataset.csv')
```

```python
# inspect the dataset
print(df.info())
print('\n')
print(df.describe())
```

```python
# view the mean, median, and mode
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        print(f'{col}:')
        print(f'  Mean:   {df[col].mean():.2f}')
        print(f'  Median: {df[col].median()}')
        print(f'  Mode:   {df[col].mode().tolist()}')

for col in df.columns:
    if not pd.api.types.is_numeric_dtype(df[col]):
        print(f'{col}:')
        print(f'  Mode:   {df[col].mode().tolist()}')
```

```python
# show the standard deviation aka how much does the values differ from the avg value.
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        print(f'{col} std: {df[col].std():.2f}')
```

```python
# show the range of values
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        col_min = df[col].min()
        col_max = df[col].max()
        col_range = round(col_max - col_min, 2)
        print(f'{col}: min = {col_min}, max = {col_max}, range = {col_range}')

print('\n=== Categorical Features: Unique Categories ===')
for col in df.columns:
    if not pd.api.types.is_numeric_dtype(df[col]):
        print(f'{col} ({df[col].nunique()} unique categories): {df[col].unique().tolist()}')
```
