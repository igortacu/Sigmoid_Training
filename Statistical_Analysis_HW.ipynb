```python
import pandas as pd
```

```python
DATASET_FILE = 'credit_risk_dataset.csv'
credit_risk_data = pd.read_csv(DATASET_FILE)
```

```python
print(credit_risk_data.info())
print('\n')
print(credit_risk_data.describe())
```

```python
for col in credit_risk_data.columns:
	if pd.api.types.is_numeric_dtype(credit_risk_data[col]):
		print(f'{col}:')
		print(f'  Mean:   {credit_risk_data[col].mean():.2f}')
		print(f'  Median: {credit_risk_data[col].median()}')
		print(f'  Mode:   {credit_risk_data[col].mode().tolist()}')

for col in credit_risk_data.columns:
	if not pd.api.types.is_numeric_dtype(credit_risk_data[col]):
		print(f'{col}:')
		print(f'  Mode:   {credit_risk_data[col].mode().tolist()}')
```

```python
for col in credit_risk_data.columns:
	if pd.api.types.is_numeric_dtype(credit_risk_data[col]):
		print(f'{col} std: {credit_risk_data[col].std():.2f}')
```

```python
for col in credit_risk_data.columns:
	if pd.api.types.is_numeric_dtype(credit_risk_data[col]):
		feature_min = credit_risk_data[col].min()
		feature_max = credit_risk_data[col].max()
		feature_range = round(feature_max - feature_min, 2)
		print(f'{col}: min = {feature_min}, max = {feature_max}, range = {feature_range}')

for col in credit_risk_data.columns:
	if not pd.api.types.is_numeric_dtype(credit_risk_data[col]):
		print(f'{col} ({credit_risk_data[col].nunique()} unique categories): {credit_risk_data[col].unique().tolist()}')
```
